```python
import sqlite3

# Path to your SQLite database file
db_path = "app.db"

# Connect to SQLite
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

# Drop the table if it exists
cursor.execute("DROP TABLE IF EXISTS annual_vacations;")
print("Table 'annual_vacations' dropped successfully (if it existed).")

# Commit changes and close connection
conn.commit()
conn.close()
```
